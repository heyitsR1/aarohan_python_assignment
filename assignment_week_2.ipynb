```python
#Question 1

li = list(input ("Enter a list of numbers").split(","))
for n in li: 
    if int(n) > 50: 
        break 
    if int(n) % 5 == 0: 
        continue
    print(n)
```

```python
#Question 2
sen_list = list(input ("Enter string: ").split (" "))
counter = 0
rev = ""

for x in sen_list: 
    counter +=1
    if counter % 2 != 0: 
        rev = rev + " " + x
        continue
    rev = rev +" "+ x[::-1]

print (rev)
```

```python
#Question 3
word_list = list(input ("Enter words: ").split (","))
count = {}
for word in word_list: 
    if word in count: 
        count[word] +=1
        continue
    count[word]=1

print (count)
```

```python
#Question 4
books = {
    'Book1': 5,
    'Book2': 6,
    'Book3': 10,
}

book_name = input ("Name of Book: ")
copies = input ("Number of copies you want to buy: ")

while not copies.isnumeric():
    print (" Please enter a valid number of books (integer) ")
    copies = input ("Number of copies you want to buy: ")

if book_name not in books: 
    print ("Book unavailable")
elif books[book_name] < copies: 
    print ("Partially available")
else: 
    print ("Available")
```

```python
#Question 6
def largest_word (sentence): 
    return max([(len(x),x) for x in sentence.split(" ")])

def main (): 
    sentence = input ("Enter a sentence: ")
    print (largest_word(sentence)[1])

main()
```

```python
#Question 7
def title_case (sentence):
    return ' '.join([x[0].upper()+x[1:] if len(x) > 1 else x.upper() for x in sentence.split(" ")])

def main (): 
    sentence = input ("Enter a sentence: ")
    print (title_case(sentence))

main()
```
