```python
def arry_function (inpu):
    out = list(filter(lambda x: x if x% 2 ==0 else None,inpu))
    sum = 0
    print ("Even Numbers:" ,out)
    for i in inpu: 
        if i not in out: 
            sum +=i
    return sum


def main():
    inpu = [10,15,22,33,47,50] 
    print ("Sum of odd is ", arry_function(inpu))

main()
```
