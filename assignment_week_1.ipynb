```python
#Question 4
password = input ("Enter Password: ")
print (password)
if (len(password) < 6 or password.isalpha()): 
    print ("Weak Password")
elif len(password) >= 6  and any(c.isdigit() for c in password) and any (c.isalpha() for c in password): 
    if len(password) >= 8 and any(c in "@#$%&" for c in password):
        print ("Strong Password")
    else: 
        print ("Moderate password")
else: 
    print ("Weak Password")
```

```python
#Question 5
s = input ("Enter a string: ").lower()
counted = []
for c in s: 
    if c not in counted: 
        print (c, " -> ", s.count(c))
        counted.append(c)
```

```python
#Question 6
s = input ("Enter a string: ")
print (s)
vowels = 0  
consonants = 0
for c in s: 
    if not c.isalnum(): 
        continue
    elif c in ['a','e','i','o','u']: 
        vowels +=1
    else: 
        consonants +=1
print ("Vowels: " , vowels , " Consonants : " , consonants)


```
